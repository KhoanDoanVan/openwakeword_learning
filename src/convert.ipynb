```python
import coremltools as ct
import onnx
from exceptiongroup import catch
```

```python
import os
```

```python
def convert_onnx_to_coreml(onnx_path, coreml_name):

    try:
        onnx_model = onnx.load(onnx_path)

    except Exception as e:
        print(f"Error Loading onnx model: {e}")


    # In general cases, Core ML Tools will au tomatic thinking input/output.
    # In complex models or has a layer not supported, you must specify more metrics for ct.convert().

    # If model have dynamic inputs/outputs, you need config ct.Shape(shape_range).
    # Otherwise, ct.convert() will set default length of ONNX is static.
    try:
        coreml_model = ct.convert(
            onnx_model,
            source="onnx",
            minimum_deployment_target=ct.target.iOS13
        )


        output_path = f"{coreml_name}.mlpackage"
        coreml_model.save(output_path)

        print("successfully converted onnx model to mlpackage!!!")

    except Exception as e:
        print(f"Error converting onnx model to coreml: {e}")
```

```python
convert_onnx_to_coreml("./resources/models/embedding_model.onnx", "EmbeddingModel")
```
